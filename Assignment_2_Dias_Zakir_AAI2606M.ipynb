Normalization rescales features to a common range (here [0, 1]) so that variables measured in different units, like area in square meters and price in millions, can be compared fairly and none of them dominates just because of its larger scale. It is especially useful for distance-based algorithms such as k-NN and k-means, for models trained with gradient descent like linear regression and neural networks, and for visualizing or combining features into a single score.

## Renaming columns and feature engineering

```python
new_columns = [
    'id',
    'date',
    'district',
    'residency',
    'address',
    'area',
    'rooms',
    'floor',
    'total_floors',
    'price',
    'year',
    'distance',
    'renovation',
    'property',
    'balcony',
    'seller'
]
```

```python
apartment_df.head()
```

```python
import datetime as dt

apartment_df['building_age'] = dt.datetime.now().year - apartment_df['year']
apartment_df.head()
```

```python
# Area category
apartment_df['area_category'] = pd.qcut(apartment_df['area'], q=4, labels=['small', 'medium', 'large', 'very_large'])
apartment_df.head()
```

```python
# Price category
apartment_df['price_category'] = pd.qcut(apartment_df['price_per_m2'], q=3, labels=['cheap', 'mid', 'expensive'])
apartment_df.head()
```

```python
apartment_df.keys()
```

```python
# Room area ratio
apartment_df['room_area_ratio'] = apartment_df['area'] / apartment_df['rooms'].replace(0,1)
apartment_df['room_area_ratio'] = apartment_df['room_area_ratio'].round(2)
apartment_df.head(3)
```

### New Features

**`area_category`**: groups apartments by total area into size classes (`small`, `medium`, `large`, `very_large`). It turns a continuous variable into interpretable segments, making it easier to compare prices and other characteristics across apartment sizes.

**`price_category`**: splits apartments into price tiers (`cheap`, `mid`, `expensive`) using quantiles, so each group contains roughly the same number of listings. It helps analyze which characteristics are typical for each market segment. *Note: since it is derived from the target variable, it is used only for exploratory analysis and is not passed to the model.*

**`room_area_ratio`**: average area per room (`total_area / rooms`), in m². It shows how spacious the apartment is: two apartments with the same total area can differ a lot in comfort depending on how many rooms the space is divided into.

**`building_age`**: age of the building in years (`current year − year built`). Older buildings may have outdated infrastructure, while newer ones usually have higher prices, so age is expected to be an important price factor.

## Correlation analysis and final conclusions

### Correlation with price

The features most strongly correlated with `price` are:

| Feature | Correlation with price |
|---|---|
| `area` | 0.85 |
| `rooms` | 0.74 |
| `price_per_m2` | 0.54 |
| `distance` | −0.24 |

Note: `price_per_m2` is derived from `price`, so its correlation is partly a result of how the feature was constructed.

**Positive correlation:** `area` (0.85): larger apartments tend to have higher prices.

**Negative correlation:** `distance` (−0.24): apartments located farther away tend to be cheaper. The relationship is weak, though.

**Correlation does not imply causation.** These coefficients only show that features change together, not that one causes the other. For example, the negative correlation between `distance` and `price` may be partly explained by other factors, such as district or building age, rather than by distance itself.

### Final Conclusions

1. **Location strongly affects price (grouping).** Yesil is the most expensive district for every number of rooms, while Baikonur is the cheapest. A 1-room apartment costs on average 29.2 million in Yesil versus 18.5 million in Baikonur, about 1.6 times more.

2. **Renovation quality is associated with higher price per m² (grouping).** Among regular apartments, designer renovation has the highest median price per m² (0.639), and rough finish has the lowest (0.503), a difference of about 27%. # Questions

## A. NumPy

**1. Array vs list.** A NumPy array stores elements of one type in a contiguous block of memory and supports element-wise math. A list can hold mixed types, stores references to objects, and `+` concatenates lists instead of adding numbers.

**2. shape, ndim, size, dtype.** `shape` is the size along each axis, e.g. `(7, 24, 3)`. `ndim` is the number of axes. `size` is the total number of elements. `dtype` is the data type of the elements, e.g. `float64`.

**3. Why arrays are faster.** Operations run in optimized compiled C code over contiguous memory of a single type, with no Python-level loop and no per-element type checks.

**4. Vectorization.** Applying an operation to a whole array at once instead of looping. Example from Task 1: `price_per_m2_array = prices / areas`.

**5. Broadcasting.** NumPy's rule for operating on arrays of different shapes by virtually stretching the smaller one. Used when combining an array with a scalar (`arr * 2`) or a row/column with a matrix (`matrix - row_means`).

**6. Indexing vs slicing.** Indexing (`arr[5]`) returns a single element. Slicing (`arr[2:5]`) returns a subarray, which is a view of the original data.

**7. `arr[10:20]`.** Returns elements at indices 10 through 19. Index 20 is not included.

**8. Last 10 elements.** `arr[-10:]`

**9. Boolean mask.** An array of `True`/`False` values (dtype `bool`), e.g. `prices > 50`. Used to select the elements where the value is `True`.

**10. `&` / `|` vs `and` / `or`.** `and`/`or` need a single True/False and fail on arrays ("truth value is ambiguous"). `&`/`|` work element-wise. Each condition must be in parentheses because `&` has higher precedence than `>`.

**11. Filter on three conditions.**
`apts[(areas > 50) & (rooms == 2) & (prices < 40)]`

**12. Mean vs median.** The mean is the arithmetic average. The median is the middle value of the sorted data. The median is more informative when there are outliers or the distribution is skewed, as with apartment prices.

**13. Standard deviation.** How far values typically spread around the mean, in the same units as the data.

**14. Variance and std.** Variance is the average squared deviation from the mean. Standard deviation is its square root.

**15. Percentile.** The value below which a given percent of observations fall. A 75th percentile price of X means 75% of apartments cost X or less.

**16. `np.argmax()`.** Returns the index (position) of the maximum value, not the value itself. Example: `prices[np.argmax(prices)]` gives the maximum price.

**17. `price_per_m2` without a loop.** Element-wise division of two arrays of equal shape is vectorized: `prices / areas`.

**18. Min-Max normalization.** `(x − min) / (max − min)`. It rescales values to the range [0, 1].

**19. If `max == min`.** The denominator is 0, so the result is NaN or inf. Handle it separately, for example by setting all values to 0.

**20. Tensor.** An array with more than two axes, a generalization of vectors (1D) and matrices (2D).

**21. Axes of T (7, 24, 3).** 7 days × 24 hours × 3 cities.

**22. `T[:, :, 1]`.** All days and hours for the city at index 1. Shape `(7, 24)`.

**23. `T[0, :, :]`.** All hours and cities for the first day. Shape `(24, 3)`.

**24. All cities at 8:00 for all days.** `T[:, 8, :]`, shape `(7, 3)`.

**25. The `axis` parameter.** The axis along which the function aggregates. That axis disappears from the result. For example, `axis=0` collapses the first dimension.

**26. One average per city.** Average over days and hours: `T.mean(axis=(0, 1))`, shape `(3,)`.

**27. Average per city per day, shape (7, 3).** Average over hours: `T.mean(axis=1)`.

**28. `reshape()` vs `transpose()`.** `reshape` changes the shape but keeps the order of elements in memory. `transpose` reorders the axes (rows become columns), which changes how the elements are arranged.

**29. Why `reshape(7, 72)` keeps the count.** 7 × 24 × 3 = 504 = 7 × 72. `reshape` only regroups the elements, and the new shape must have the same total size.

**30. `np.array_equal()` / `np.allclose()`.** Checking that two arrays match, for example that a vectorized result equals a loop result. `array_equal` requires exact equality. `allclose` allows small floating-point differences.

**31. Correlation coefficient.** The strength and direction of a linear relationship between two variables.

**32. Range.** From −1 to 1.

**33. +1, −1, 0.** +1 is a strong positive linear relationship (both grow together). −1 is a strong negative one (one grows, the other falls). 0 means no linear relationship, though a non-linear one may still exist.

**34. Correlation ≠ causation.** Both variables may depend on a third factor (a confounder), the relationship may be coincidental, or the direction of influence may be reversed.

## B. Pandas

**1. Series vs DataFrame.** A Series is a one-dimensional labeled array, like a single column. A DataFrame is a two-dimensional table of columns, where each column is a Series.

**2. head/tail/shape/columns/dtypes.** `head()`/`tail()` show the first or last rows. `shape` gives (rows, columns). `columns` lists the column names. `dtypes` gives each column's type.

**3. `df.info()`.** Number of rows, column names, non-null count per column, dtypes, and memory usage.

**4. `describe()` vs `info()`.** `info()` describes the structure: types and missing values. `describe()` gives statistics: count, mean, std, min, quartiles, max.

**5. `pd.concat()` and `ignore_index`.** Combines DataFrames by stacking rows or columns. `ignore_index=True` creates a new 0..n−1 index; otherwise the original indices are kept and can repeat.

**6. Fully duplicated row.** A row identical to another row in every column.

**7. `duplicated()` vs `drop_duplicates()`.** `duplicated()` returns a boolean mask marking repeated rows. `drop_duplicates()` returns the DataFrame without them.

**8. Why not always remove duplicates automatically.** Identical rows can be legitimate, such as two identical apartments in one building. It's also important to decide which copy to keep, and duplicates by key may hold conflicting values.

**9. Duplicates on selected columns.** `df.duplicated(subset=['id'])`, or with several columns in the list.

**10. Missing values per column.** `df.isna().sum()`

**11. Percentage of missing values.** `df.isna().mean() * 100`

**12. `dropna()` vs `fillna()`.** `dropna()` removes rows or columns with missing values. `fillna()` replaces missing values with a chosen value.

**13. Mean vs median for imputation.** Use the mean for roughly symmetric data without outliers. Use the median for skewed data or data with outliers, such as price or area.

**14. Mode.** For categorical features, filling with the most frequent category is reasonable when the share of missing values is small.

**15. Group-based imputation.** Filling missing values with a group statistic, e.g. the median area within the same number of rooms. It is more accurate because it accounts for differences between groups.

**16. `pd.to_numeric(errors="coerce")`.** Converts a column to numbers. Values that can't be converted become NaN instead of raising an error.

**17. `pd.to_datetime()`.** Converts strings to datetime, which enables date arithmetic, extracting year or month, sorting, and time-based filtering.

**18. Check NaN after conversion.** With `errors="coerce"`, invalid values silently become NaN, so new missing values appear and must be handled.

**19. Renaming columns.** `df.rename(columns={'old': 'new', 'old2': 'new2'})`

**20. Feature engineering.** Creating new features from existing ones to make patterns more visible. In Task 3: `price_per_m2`, `area_category`, `price_category`, `room_area_ratio`, `building_age`.

**21. Zero area before `price_per_m2`.** Dividing by zero gives inf, which distorts means, sorting, and plots. Zero area is also a data error in itself.

**22. `loc[]` vs `iloc[]`.** `loc` selects by labels (index names, column names, boolean masks). `iloc` selects by integer positions.

**23. Filter on several conditions.**
`df[(df['rooms'] == 2) & (df['price'] < 40)]`, each condition in parentheses, joined with `&` / `|`.

**24. `sort_values()`.** Sorts rows by one or more columns, ascending by default (`ascending=False` for descending).

**25. `groupby()` and split-apply-combine.** Split the data into groups by a key, apply a function to each group (mean, count, etc.), and combine the results into one table.

**26. `agg()` vs a single `mean()`.** `agg()` computes several statistics at once (count, mean, median, min, max), and different ones per column, giving a fuller picture of each group.

**27. `pivot_table` vs `groupby`.** A pivot table puts one grouping key on rows and another on columns, producing a cross-table that's easy to read. `groupby` gives a long table with a multi-index.

**28. `count` in an aggregated table.** The number of rows in each group. It shows how reliable each group's statistic is.

**29. Small-count groups.** With only a few observations, the mean depends heavily on individual values and outliers, so the result may be random rather than a real pattern.

**30. Chart types.** A bar chart compares a value across categories (mean price by rooms). A scatter plot shows the relationship between two numeric variables. A boxplot shows a distribution: median, quartiles, spread, and outliers, often across groups. A histogram shows the frequency distribution of one numeric variable.

**65. Scatter of price vs area.** A positive, roughly linear trend, a spread that widens with area, clusters by number of rooms, and possible outliers.

**31. Boxplot and outliers.** The box spans Q1 to Q3 (the middle 50%), the line inside is the median, and the whiskers extend up to 1.5 × IQR. Points beyond the whiskers are outliers: values unusually far from the rest.

**32. Histogram.** The shape of a distribution: where values concentrate, skewness, spread, and gaps.

**33. Correlation matrix.** A table of pairwise correlation coefficients between all numeric columns.

**34. Diagonal equals 1.** Each feature is perfectly correlated with itself.

**35. Pearson on text columns.** No. Pearson requires numeric values. Text must first be encoded, and even then the correlation is only meaningful for ordered categories.

**36. Correlation ≠ causation.** Same as Q34: confounders, coincidence, reversed direction. Proving causation requires experiments or causal analysis.

**37. Save CSV without the index.** `df.to_csv('file.csv', index=False)`

**38. Verify the saved CSV.** Read it back with `pd.read_csv()` and compare `shape`, `dtypes`, `head()`, and missing values with the original, or use `df.equals()`.

**39. merge vs join vs concat.** `merge` joins tables on shared columns, like SQL JOIN. `join` joins on the index. `concat` stacks tables along rows or columns without matching keys.

**40. `groupby` on a column with NaN.** Rows with NaN in the key are excluded by default. `groupby(..., dropna=False)` keeps them as a separate group.

**41. Top/bottom 5 by a feature.** `df.nlargest(5, 'col')` / `df.nsmallest(5, 'col')`, or `sort_values('col').head(5)`.
